**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
injury_columns = [
    'Total.Fatal.Injuries',
    'Total.Serious.Injuries',
    'Total.Minor.Injuries',
    'Total.Uninjured',
    'Aircraft.damage'
]
df[injury_columns].isnull().sum()
```

```python
df[injury_columns].describe(include='all')
```

```python
# Fill missing injury counts with 0
injury_count_columns = [
    'Total.Fatal.Injuries',
    'Total.Serious.Injuries',
    'Total.Minor.Injuries',
    'Total.Uninjured'
]

df[injury_count_columns] = df[injury_count_columns].fillna(0)
```

```python
#   Total number of passengers on board
df["Total.Passengers"]=df[injury_count_columns].sum(axis=1)

# Calculate fatal and serious injuries
df['Fatal.Serious.Injuries'] = (
    df['Total.Fatal.Injuries'] + df['Total.Serious.Injuries']
)

# Calculate the proportion of people with fatal or serious injuries
df['Serious.Fatal.Injury.Rate'] = (
    df['Fatal.Serious.Injuries'] / df['Total.Passengers'].replace(0, float('nan'))
)
df['Serious.Fatal.Injury.Rate']
```
